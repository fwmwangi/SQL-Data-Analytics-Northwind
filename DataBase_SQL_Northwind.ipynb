##  Part 1: Quick Reference SQL Cheat Sheet

| SQL Concept | Syntax Example | What It Does |
| :--- | :--- | :--- |
| **SELECT & Alias** | `SELECT column_name AS ID` | Renames columns in output for cleaner reporting. |
| **CASE WHEN** | `CASE WHEN jobTitle LIKE '%VP%' THEN 'Executive' ELSE 'Not Executive' END` | Adds conditional logic/binning directly in the database query. |
| **String SUBSTR** | `SUBSTR(jobTitle, 1, 2)` | Extracts specific characters (start position, length). |
| **String LENGTH** | `LENGTH(lastName)` | Counts total characters in a text field. |
| **Math ROUND** | `ROUND(priceEach * quantityOrdered)` | Rounds numerical results to prevent floating-point decimals. |
| **Date STRFTIME** | `STRFTIME('%Y', orderDate)` | Extracts Day (`%d`), Month (`%m`), or Year (`%Y`) from timestamps. |

## Part 2: CodeGrade Step-by-Step Solutions for `Database_SQL_Northwind.ipynb`

### Step 1: Establish Database Connection

```python
# step1
import sqlite3
import pandas as pd

conn = sqlite3.connect("data.sqlite")
```

### Step 2: Select Specific Columns

```python
# step2
df_name_number = pd.read_sql("""
    SELECT employeeNumber, lastName 
    FROM employees
""", conn)

# Display Output for visual presentation
df_name_number.head()
```

### Step 3: Reverse Column Order

```python
# CodeGrade step3
df_reverse = pd.read_sql("""
    SELECT lastName, employeeNumber 
    FROM employees
""", conn)

df_reverse
```

### Step 4: Column Aliasing

```python
df_alias = pd.read_sql("""
    SELECT lastName, employeeNumber AS ID 
    FROM employees
""", conn)

df_alias.head(5)
```

### Step 5: Bin Roles with Conditional Logic (`CASE`)

```python
# CodeGrade step5
df_executive = pd.read_sql("""
    SELECT *,
        CASE 
            WHEN jobTitle IN ('President', 'VP Sales', 'VP Marketing') THEN 'Executive'
            ELSE 'Not Executive'
        END AS role
    FROM employees
""", conn)

df_executive.tail(5)
```

### Step 6: String Functions — Measure String Length

```python
# CodeGrade step6
df_name_length = pd.read_sql("""
    SELECT LENGTH(lastName) AS name_length 
    FROM employees
""", conn)

df_name_length.head()
```

### Step 7: String Functions — Extract Substring

```python
# CodeGrade step7
df_short_title = pd.read_sql("""
    SELECT SUBSTR(jobTitle, 1, 2) AS short_title 
    FROM employees
""", conn)

df_short_title.head()
```

### Step 8: Numeric Aggregation — Total Revenue Calculation

```python
# CodeGrade step8
sum_total_price = pd.read_sql("""
    SELECT CAST (round(priceEach * quantityOrdered) AS INTEGER) total_price
    FROM orderDetails
""", conn).sum()

sum_total_price
```

### Step 9: Date Formatting & Extraction

```python
# CodeGrade step9
df_day_month_year = pd.read_sql("""
    SELECT 
        orderDate,
          strftime("%d", orderDate) Day,
               CASE strftime("%m", orderDate)
               WHEN '01' THEN 'January'
               WHEN '02' THEN 'February'
               WHEN '03' THEN 'March'
               WHEN '04' THEN 'April'
               WHEN '05' THEN 'May'
               WHEN '06' THEN 'June'
               WHEN '07' THEN 'July'
               WHEN '08' THEN 'August'
               WHEN '09' THEN 'September'
               WHEN '10' THEN 'October'
               WHEN '11' THEN 'November'
               WHEN '12' THEN 'December'
           END Month,
          strftime("%Y", orderDate) Year
          
   FROM orders
   """, conn)

# Display Output for visual presentation
df_day_month_year.head()
```

### Close Connection

```python
conn.close()
```
